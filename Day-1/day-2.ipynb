```python
import numpy as np
```

```python
#swapaxes
original = np.arange(30).reshape(2,3,5)
print(original)
swapped = np.swapaxes(original,1,2)  #indexes of deimensions
print(swapped)
```

```python
#squeeze
arr=np.arange(20).reshape(1,5,4)
print(arr)
print(arr.shape)
temp =arr.squeeze() #remove 1 dimensions
print(temp,temp.shape)
```
